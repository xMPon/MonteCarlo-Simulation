# Monte Carlo Pi Estimation

This notebook runs a simple Monte Carlo simulation to estimate $\pi$.

```python
# Put the project root on sys.path so `simulations` is importable.
# The notebook lives in notebooks/, but some editors start the kernel at the
# repo root instead, so locate the root by looking for simulations.py.
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / "simulations.py").exists():
    project_root = project_root.parent
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

print(f"Project root: {project_root}")
```

```python
from simulations import estimate_pi

# Run the Monte Carlo simulation to estimate pi
samples = 200_000  # Number of random points
seed = 42  # Random seed for reproducibility
pi_hat = estimate_pi(samples=samples, seed=seed)
pi_hat
```

```python
# Calculate the absolute error from the true value of pi
import math

abs(pi_hat - math.pi)
```
